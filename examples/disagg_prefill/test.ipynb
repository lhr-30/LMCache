```python
import requests
import json
import sys

# vLLM OpenAI-compatible API 地址
url = "http://localhost:9100/v1/completions"  # 换成你 decoder 端口，比如 7100/7200

# 请求体
payload = {
    "model": "Qwen/Qwen2.5-7B",  # 模型名字要和 serve 时一致
    "prompt": "what is the capital of France",
    "max_tokens": 16,
    "temperature": 0,
}

# 发送 POST 请求
try:
    resp = requests.post(url, json=payload, timeout=30)
except requests.RequestException as e:
    print("Request failed:", repr(e))
    sys.exit(1)

print("Status:", resp.status_code)
print("Content-Type:", resp.headers.get("Content-Type"))
print("URL hit:", resp.request.method, resp.request.url)

# 打印前 800 字，防止太长
body_preview = resp.text[:800] if resp.text else ""
print("Raw body preview:\n", body_preview)

# 若不是 2xx，直接退出（通常不是 JSON）
if not resp.ok:
    print("\nNon-2xx response; likely不是JSON。请检查上面的原始正文（可能是HTML报错/404/后端栈追踪）。")
    sys.exit(1)

# 尝试解析 JSON
try:
    data = resp.json()
    print("\nParsed JSON:\n", json.dumps(data, indent=2))
except Exception as e:
    print("\nJSON 解析失败：", repr(e))
    print("提示：如果 Content-Type 不是 application/json，或 body 是空/HTML，就会失败。")
```
